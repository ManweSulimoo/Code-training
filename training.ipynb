```python
def bubble_sort(L:list[float])->list:

    """
    Sorts a list in ascending order using bubble sort
    Parameter(s):
    L (list[float]): Input list
    Return(s):
    list: Sorted list
    """

    if not isinstance(L, list):
        raise TypeError("Input should be a list")
    
    bridge = 0

    while not all(L[i]<L[i+1] for i in range(len(L)-1)):
        for i in range(len(L)-1):
            if L[i]>L[i+1]:
                bridge = L[i]
                L[i] = L[i+1]
                L[i+1] = bridge
                
    return L


def bubble_sort(L: list[float]) -> list:
    """
    Sorts a list in ascending order using bubble sort
    Parameter(s):
    L (list[float]): Input list
    Return(s):
    list: Sorted list
    """
    if not isinstance(L, list):
        raise TypeError("Input should be a list")
    
    n = len(L)
    for i in range(n):
        swapped = False
        for j in range(n - 1 - i):
            if L[j] > L[j+1]:
                L[j], L[j+1] = L[j+1], L[j]
                swapped = True
        if not swapped:
            break  # déjà trié, on arrête
    return L


bubble_sort([5,2,4,1])
```

```python
def string_compression(S:str)->str:
    """ 
    Returns the compressed format of a string.

    Parameter(s):
    S (str): Input string

    Return(s):
    Returning the compressed format of the input string
    """

    if not isinstance(S, str):
        raise TypeError("Input should be a string")
    
    format_str = ''
    current_char = ''

    for i,s in enumerate(S):
        if not current_char:
            current_char = s
            count = 1
            continue
        if s == current_char:
            count += 1
        if s != current_char:
            format_str += current_char + f'{count}'
            current_char = s
            count = 1
        
        if i == len(S)-1:
            format_str += current_char + f'{count}'
            return format_str
        
```

```python
def factor_recursive(n:int)->int:
    """ 
    Returns the factorial of an integer

    Parameter(s):
    n (int): Input integer

    Return(s):
    Returning the factorial of the input integer
    """

    if not isinstance(n, int) or not n > 0:
        raise TypeError("Input should a positive integer")
    

    if n ==1 or n ==2:
        facto = n
    else:
        facto = n * factor_recursive(n-1)

    return facto
```

```python
import numpy as np

def gradient_descent(X: list, y: float, epochs: int = 1000, lr: float = 1e-1):
    """
    Trains a linear regression model via gradient descent
    Parameter(s):
    X (list): Input features
    y (float): Target value
    epochs (int): Number of training iterations
    lr (float): Learning rate
    Return(s):
    tuple: Trained weights W and bias b
    """
    if not isinstance(X, list):
        raise TypeError("Input values should be a list")
    if not isinstance(y, float):
        raise TypeError("Target value should be a float")
    
    # Initialisation
    W = [np.random.uniform(0., 1.) for _ in X]
    b = 0.0
    
    for epoch in range(epochs):
        # Forward pass
        y_pred = b
        for x, w in zip(X, W):
            y_pred += x * w
        
        # Loss (MSE)
        L = (y_pred - y) ** 2
        
        # Backpropagation
        dL_dW = [2 * (y_pred - y) * x for x in X]
        dL_db = 2 * (y_pred - y)
        
        # Update
        W = [w - lr * dw for w, dw in zip(W, dL_dW)]
        b = b - lr * dL_db
        
        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {L:.4f}")
    
    return (W, b)

gradient_descent([2,3], 1.)
```

```python
import math as m
def logistic_regression(X: list, y: float, epochs: int = 1000, lr: float = 1e-1):
    """ 
    Trains a logistic regression by gradient descent

    Parameter(s):
    X (list): Input features
    y (float): Target value
    epochs (int): Number of training iterations
    lr (float): Learning rate

    Return(s):
    tuple: Updates weights and bias.
    """

    if not isinstance(X, list) or not isinstance(y, float):
        raise TypeError("Input features should be given as a list, and the target value should be a float")
    
    ## Initialization
    W = [np.random.uniform(0., 1.) for x in X]
    b = 0.0

    ## Sigmoid function
    def sigmoid(z):
        return 1/(1 + m.exp(-z))
    
    
    for epoch in range(epochs):
        # Forward pass + sigmoid
        z = b
        for x,w in zip(X,W):
            z += x * w
            #y_pred = sigmoid(z)
            y_pred = min(max(sigmoid(z), 1e-7), 1 - 1e-7)  # clipping

        # Loss (Cross-entropy)
        CE = -(y * m.log(y_pred) + (1-y) * m.log(1-y_pred))

        # Back propagation
        dL_dW = [(y_pred - y) * x for x in X]
        dL_db = (y_pred - y)

        # Update
        W = [w - lr * dw for (w,dw) in zip(W, dL_dW)]
        b = b - lr * dL_db

        if epoch % 100 == 0:
            print(f"Epoch: {epoch}, Loss: {CE}")

    return (W,b)


logistic_regression([2.71, 3.33], 0.86)
```

```python
def sample_split(X: np.ndarray, y: np.ndarray, 
                 train_size: float = 0.8, 
                 test_size: float = 0.1, 
                 val_size: float = 0.1):
    """
    Returns training, test and validation splits from input data
    Parameter(s):
    X (np.ndarray): Input features
    y (np.ndarray): Labels
    train_size (float): Proportion for training
    test_size (float): Proportion for test
    val_size (float): Proportion for validation
    Return(s):
    tuple: X_train, y_train, X_test, y_test, X_val, y_val
    """
    if not isinstance(X, np.ndarray) or not isinstance(y, np.ndarray):
        raise TypeError("Features and labels should be numpy arrays")
    if not all(isinstance(s, float) for s in [train_size, test_size, val_size]):
        raise ValueError("Split sizes should be floats")
    if not round(train_size + test_size + val_size, 5) == 1.0:
        raise ValueError("Split sizes should sum to 1.0")
    
    n = X.shape[0]
    
    # Shuffle des indices
    indices = np.random.permutation(n)
    
    # Calcul des seuils
    train_end = round(train_size * n)
    test_end  = train_end + round(test_size * n)
    
    train_idx = indices[:train_end]
    test_idx  = indices[train_end:test_end]
    val_idx   = indices[test_end:]
    
    return (X[train_idx], y[train_idx],
            X[test_idx],  y[test_idx],
            X[val_idx],   y[val_idx])

X = np.array([[2,3,7], [3,4,8], [6,8,9], [5,9,1], [8,7,6], [9,8,1]])
y = np.array([0,1,1,2,0,1])
sample_split(X, y, train_size=0.7, test_size=0.2, val_size=0.1)
```

```python
import math as m
def metrics(y_pred : np.ndarray, y_true : np.ndarray, method : str = "regression"):

    """
    Parameter(s):
    y_pred (np.ndarray): Predictions from the model
    y_true (np.ndarray): True values (or labels for classif)
    method (str): Type of problem solved by the model, either 'regression' or 'classification'

    Return(s):
    tuple: MSE, MAE, RMSE, accuracy, F1
    """

    if not isinstance(y_pred, np.ndarray) or not isinstance(y_true, np.ndarray):
        raise TypeError("Value inputs should be arrays")
    if not isinstance(method, str) or method not in ["regression", "classification"]:
        raise ValueError("Method should be a string either 'regression' or 'classification'")
    if not len(y_pred)==len(y_true):
        raise ValueError("Input arrays should be the same size")
    
    n = len(y_pred)

    ### MSE, MAE and RMSE for regression
    if method == "regression":
        MSE = (1/n) * sum((y_pred-y_true)**2)
        MAE = (1/n) * sum(abs(y_pred-y_true))
        RMSE = m.sqrt(MSE)
        return MSE, MAE, RMSE
    

    ### Accuracy, precision, recall and F1 score
    else:
        TP = np.count_nonzero((y_true==1) & (y_pred==1))
        TN = np.count_nonzero((y_true==0) & (y_pred==0))
        FP = np.count_nonzero((y_true==0) & (y_pred==1))
        FN = np.count_nonzero((y_true==1) & (y_pred==0))

        accuracy = (TP+TN) / (TP+TN+FP+FN)

        # Si TP+FP == 0 ou TP+FN == 0 → division par zéro
        precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
        recall    = TP / (TP + FN) if (TP + FN) > 0 else 0.0

        F1 = 2 * (precision*recall) / (precision+recall) if (precision+recall) > 0 else 0.0

                
        return accuracy, precision, recall, F1


y_true_regression = np.array([1.2, 3.2, 4.5, 9.3])
y_pred_regression = np.array([1.1, 3.5, 4.6, 8.8])

y_true = np.array([0, 1, 1, 0])
y_pred = np.array([1, 1, 0, 0])

metrics(y_true, y_pred, 'classification')
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class CNN_images(nn.Module):
    def __init__(self, in_channels:int, num_classes:int):
        """
        CNN for image classification (e.g. CIFAR10)
        Parameter(s):
        in_channels (int): Number of input channels (3 for RGB)
        num_classes (int): Number of output classes
        """
        super().__init__()
        #1st convolution layer
        self.conv2d_1 = nn.Conv2d(in_channels=in_channels, out_channels=16, kernel_size=3, padding=1)
        #2nd convolution layer
        self.conv2d_2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)

        #1 maxpool layer
        self.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Adptative pool to adapt to different images sizes
        self.adaptive_pool = nn.AdaptiveAvgPool2d((8, 8))  # force toujours 8×8

        #2nd maxpool layer
        self.linear = nn.Linear(32*8*8, num_classes)

    def forward(self, x):

        """
        Define the forward pass of the neural network.

        Parameters:
           x (torch.tensor): Input tensor.

        Returns:
            torch.Tensor
                The output tensor after passing through the network.
       """

        x = F.relu(self.conv2d_1(x))
        x = self.maxpool(x)
        x = F.relu(self.conv2d_2(x))
        x = self.maxpool(x)
        x = self.adaptive_pool(x)
        x = x.reshape(x.shape[0], -1)
        x = self.linear(x)

        return x

class CNN_spectra(nn.Module):
    def __init__(self, in_channels:int, num_classes:int):
        """
        1D CNN for spectra classification (e.g. FSRQ / BL Lacs / BCUs)
        Parameter(s):
        in_channels (int): Number of input channels (1 for single flux array)
        num_classes (int): Number of output classes
        """
        super().__init__()
        #1st convolution layer
        self.conv1d_1 = nn.Conv1d(in_channels=in_channels, out_channels=8, kernel_size=3, padding=1)
        #2nd convolution layer
        self.conv1d_2 = nn.Conv1d(in_channels=8, out_channels=16, kernel_size=3, padding=1)

        #1 maxpool layer
        self.maxpool = nn.MaxPool1d(kernel_size=2, stride=2)

        # Adaptative pool to adapt to any spectrum length
        self.adaptive_pool = nn.AdaptiveAvgPool1d(32)  # robuste à toute longueur de spectre

        #2nd maxpool layer
        self.linear = nn.Linear(16*32, num_classes)

    def forward(self, x):

        x = F.relu(self.conv1d_1(x))
        x = self.maxpool(x)
        x = F.relu(self.conv1d_2(x))
        x = self.maxpool(x)
        x = self.adaptive_pool(x)
        x = x.reshape(x.shape[0], -1)
        x = self.linear(x)

        return x

device = "cuda" if torch.cuda.is_available() else "mps" #or 'mps' for macOS
modelA = CNN_images(in_channels=3, num_classes=10).to(device)
print(modelA)

modelB = CNN_spectra(in_channels=1, num_classes=3).to(device) #ex FSRQ, BL Lacs and BCUs
print(modelB)
```

```python
from torch.utils.data import DataLoader, TensorDataset
import random

class Stacked_LSTM(nn.Module):
    def __init__(self, input_size: int, num_hidden: int = 64, num_layers: int = 2):
        """
        Stacked LSTM for light-curve binary classification (e.g. FSRQ vs BL Lac)
        Parameter(s):
        input_size (int): Number of features per timestep
        num_hidden (int): Hidden state size
        num_layers (int): Number of stacked LSTM layers
        """
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=num_hidden,
            num_layers=num_layers,
            batch_first=True,
            dropout=0.2
        )
        self.linear = nn.Linear(num_hidden, 1)  # binaire → 1 sortie

    def forward(self, x):
        output, (h_n, c_n) = self.lstm(x)
        x = self.linear(h_n[-1])  # dernier hidden state → Linear
        return x


def create_LCs(num_LC: int = 100, seq_len: int = 64):
    """
    Creates fictive light-curves with binary labels
    Parameter(s):
    num_LC (int): Number of light-curves
    seq_len (int): Length of each light-curve
    Return(s):
    tuple: X_tensor (num_LC, seq_len, 1), y_tensor (num_LC,)
    """
    X = np.random.uniform(0, 1, size=(num_LC, seq_len, 1))  # (batch, seq, features)
    y = np.random.randint(0, 2, size=(num_LC,))

    X_tensor = torch.tensor(X, dtype=torch.float32)
    y_tensor = torch.tensor(y, dtype=torch.float32)

    return X_tensor, y_tensor


# Setup
device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"

X_tensor, y_tensor = create_LCs()
loader = DataLoader(TensorDataset(X_tensor, y_tensor), batch_size=32, shuffle=True)

# Modèle, loss, optimizer
model_LSTM = Stacked_LSTM(input_size=1).to(device)
loss_fn = nn.BCEWithLogitsLoss()  # sigmoid + BCE — plus stable
optimizer = torch.optim.Adam(model_LSTM.parameters(), lr=1e-3)

# Boucle d'entraînement
loss_history = []
for epoch in range(10):
    model_LSTM.train()  # mode train (active dropout)
    epoch_loss = 0.0

    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        output = model_LSTM(X_batch).squeeze(-1)  # (batch, 1) → (batch,)
        loss = loss_fn(output, y_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(loader)
    loss_history.append(avg_loss)
    print(f"Epoch {epoch+1}/10 — Loss: {avg_loss:.4f}")

# Classification
model_LSTM.eval()
with torch.no_grad():
    X_tensor = X_tensor.to(device)
    logits = model_LSTM(X_tensor).squeeze(-1)   # scores bruts
    probs = torch.sigmoid(logits)                # → probabilités
    preds = (probs > 0.5).int()                  # → classes 0 ou 1
    print(f"Predictions: {preds[:10]}")
    print(f"True labels: {y_tensor[:10].int()}")
```

### Transformer from scratch

#### Input embeddings

Transformers operate on sequences of embeddings. Therefore before the bottom encoder each input token (word or subword) is converted into a vector using embedding layers. These embeddings capture the semantic meaning of the tokens and convert them into numerical vectors.

All the encoders receive a list of vectors (e.g each of size 512). In the bottom encoder that would be the word embeddings, but in other subsequent encoders, it would be the output of the encoder that is directly below them.

```python
class InputEmbeddings(nn.Module): #class is defined as a subclass of PyTorch's nn.Module
    def __init__(self, vocab_size: int, d_model: int):
        """
        Input embeddings to convert tokens into vectors
        Parameter(s):
        vocab_size(int): Source vocabulary size
        d_model(int): Dimensionnality of the model's embeddings
        """
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, d_model)
        self.scale = math.sqrt(d_model) #??

    def forward(self, x):
        return self.embedding(x) * self.scale
```

#### Positional encoding

Since Transformers do not have a recurrence mechanism like RNNs, they use positional encodings added to the input embeddings to provide information about the position of each token in the sequence. This allows them to understand the position of each word within the sentence.

To do so, one method suggested by researchers consists in employing a combination of various sine and cosine functions to create positional vectors, enabling the use of this positional encoder for sentences of any length.

In this approach, each dimension is represented by unique frequencies and offsets of the wave, with the values ranging from -1 to 1, effectively representing each position.

```python
class PositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_seq_length: int):
        """ 
        Returns the input embeddings enriched (added) with positional embeddings to capture the position of each token in the sequence.
        Parameter(s)
        d_model(int): Dimension of the model's input
        max_seq_length(int): The maximum length of the sequence for which positional encodings are pre-computed.
        """
        super(PositionalEncoding, self).__init__()

        pe = torch.zeros(max_seq_length, d_model) #A tensor filled with zeros, which will be populated with positional encodings.
        position = torch.arange(0, max_seq_length, dtype=torch.float).unsqueeze(1) #A tensor containing the position indices for each position in the sequence.
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * -(math.log(10000.0) / d_model)) #A term used to scale the position indices in a specific way.

        pe[:, 0::2] = torch.sin(position*div_term) # Sine function applied to the even indices of pe -> [:, start=0:end:step=2] means we take all elements in the first dimension and in the 2nd dim we take all elemnts from the first to the last with 2 step so all even numbers
        pe[:, 1::2] = torch.sin(position*div_term) # Cosine function applied to the odd indeces of pe

        self.register_buffer('pe', pe.unsqueeze(0)) # Finally, pe is registered as a buffer, which means it will be part of the module's state but will not be considered a trainable parameter. unsqueez to extend dimension 0 (batch). 

    def forward(self, x): #The forward method simply adds the positional encodings to the input x.
        return x + self.pe[:, :x.size(1)] #It uses the first x.size(1) elements of pe to ensure that the positional encodings match the actual sequence length of x.
```

In summary:

The PositionalEncoding class adds information about the position of tokens within the sequence. 

Since the transformer model lacks inherent knowledge of the order of tokens (due to its self-attention mechanism), this class helps the model to consider the position of tokens in the sequence. 

The sinusoidal functions used are chosen to allow the model to easily learn to attend to relative positions, as they produce a unique and smooth encoding for each position in the sequence.

#### MultiHeadAttention

The multi-head attention mechanism computes the attention between each pair of positions in a sequence. It consists of multiple “attention heads” that capture different aspects of the input sequence.

Typically, the attention mechanism can be decomposed into the following steps:

1) The input embeddings are passed 3 times since they can be seen as query (q) [this is what the word is looking for], key (k) [this is what the word offers to others] and value (v) [this is the actual information the word holds]. In the bottom encoder the it will correspond to 3 times the same tensor (pure self-attention) but in cross-attention (encoder-decoder), query can come from the decoder while key and value from the encoder. 

2) Linear projection are applied to the input embeddings (q, k, v) of size (batch, seq_len, d_model). The last dimension d_model is also splitted into -> num_head * dim_head (batch, seq_len, num_head, dim_head) for multi-head attention. This results into 3 matrixes (Q, K, V).

3) The dot product attention is performed between the query matrix (Q) and the value matrix (V) so between all the tokens in position query and all the tokens in position value simultaneously by each head.

4) The heads and their information are recombined into the original embedding shape (batch, seq_len, d_model) and the result is returned.

```python
class MultiHeadAttention(nn.Module):
    def __init__(self, d_model: int, num_heads: int):
        """ 
        Performs a multihead attention on the input vector embeddings and returns the concatenate information captured by all the head as a vector with the same shape as the input emebddigns.
        Parameter(s):
        d_model(int): Dimensionnality of the model's input embeddings.
        num_heads(int): Number of heads for the attention mecanism.
        """

        super(MultiHeadAttention, self).__init__()
        # Ensure that the model dimension (d_model) is divisible by the number of heads
        assert d_model % num_heads == 0, "d_model must be divisible by num_heads"

        # Initialize dimensions
        self.d_model = d_model # Model's dimension
        self.num_heads = num_heads # Number of attention heads
        self.d_k = d_model // num_heads # Dimension of each head's key, query, and value (each head will calcualate its attention to a portion of the input embeddings corresponding to the sub diemnsion head_dim)

        # Linear layers for transforming inputs into matrixes
        self.W_q = nn.Linear(d_model, d_model) # Query transformation
        self.W_k = nn.Linear(d_model, d_model) # Key transformation
        self.W_v = nn.Linear(d_model, d_model) # Value transformation
        self.W_o = nn.Linear(d_model, d_model) # Output transformation

    #Here, the attention scores are calculated by taking the dot product of queries (Q) and keys (K), and then scaling by the square root of the key dimension (d_k).
    def scaled_dot_product_attention(self, Q, K, V, mask=None):
        attn_scores = torch.matmul(Q, K.transpose(-2,-1)) / math.sqrt(self.d_k) #Note the transpose of the last and penultimate dimensions of K to align the dimension of the two matrices. Indeed the matmul operation ignore the batch dimensions (here batch and head_dim) and so to multiply Q and K we need that their dimensions (m, n) @ (p, q) match so that n == q (linear algebra rule) -> this produces a matrix of shape (batch, num_heads, seq_len, seq_len). And so with the right matrix dimensions, we calculate for each pair of tokens in position query and key (t, s), the score of compatibility between the question of token t and the offer of token s. Geometrically this score is high when the projection of both vectors are aligned in the sub-space of dimension head dim which happens when t and s are pertinent for each other in the context of the sentence. Note also that each head calculates this attention mecanism (question/offer/content)in its own sub-space which is essential to focus on multiple aspect of the same sentence at the same time.
        # we divide by sqrt(self.dk) to prevent the scores to explode in magnitude when head_dim is large.

        # Apply mask if provided (useful for preventing attention to certain parts like padding) -> notably for the decoder part where we need to make sure that it has not seen the unformation of the target
        if mask is not None:
            attn_scores = attn_scores.masked_fill(mask == 0, -1e9) #At the positions where mask = 0, it replaces the score by -1e9, which after the softmax gives e^(-1e9) close to 0 -> these positions receive an attention weight almost null and they are subsequently invisible for the the computation of the weighted output.

        # Softmax is applied to obtain attention probabilities
        attn_probs = torch.softmax(attn_scores, dim=-1)

        # Multiply by values to obtain the final output
        output = torch.matmul(attn_probs, V)
        return output

    def split_heads(self, x):
        # Reshape the input to have num_heads for multi-head attention
        batch_size, seq_length, d_model = x.size()
        return x.view(batch_size, seq_length, self.num_heads, self.d_k).transpose(1, 2) # Here num_head becomes a dimension of type batch enabling pytorch to broadcast automatically the dimensions batch and num_head -> This transforms the dimension num_head into a dimension of type batch enabling the model to subsequently (in the method scaled_dot_product_attention) simulatenously (in parallel) calculate the attention of each head via one matrix operation between all tokens in position query and key -> this is the key of the attention mecanism.

    def combine_heads(self, x):
        # Combine the multiple heads back to original shape
        batch_size, _, seq_length, d_k = x.size()
        return x.transpose(1,2).contiguous().view(batch_size, seq_length, self.d_model)

    def forward(self, q, k, v, mask=None):
        # Apply linear transformations and split heads
        Q, K, V = map(self.split_heads, [self.W_q(q), self.W_k(k), self.W_v(v)])


        # Perform scaled dot-product attention
        attn_output = self.scaled_dot_product_attention(Q, K, V, mask)

        # Combine heads and apply output transformation
        output = self.W_o(self.combine_heads(attn_output))

        return output
```

#### Position-wise feed forward networks

The PositionWiseFeedForward class defines a position-wise feed-forward neural network that consists of two linear layers with a ReLU activation function in between. In the context of transformer models, this feed-forward network is applied to each position separately and identically. It helps in transforming the features learned by the attention mechanisms within the transformer, acting as an additional processing step for the attention outputs.

```python
class PositionWiseFeedForward(nn.Module):
    def __init__(self, d_model: int, d_ff: int):
        """ 
        Feed Forward NN which applies transformations to enrich the tokens reprensations learned during the multi-head attention process.
        Parameter(s):
        d_model(int): Dimensionnality of the model's input and output.
        d_ff(int): Dimensionality of the inner layer in the feed-forward network.
        """
        super(PositionWiseFeedForward, self).__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.fc2(self.relu(self.fc1(x))) #The output of fc1 is passed through a ReLU activation function. ReLU replaces all negative values with zeros, introducing non-linearity into the model.
```

#### Building the encoder blocks

![Encoder_architecture.png](Encoder_architecture.png)

**Architecture:** The EncoderLayer class defines a single layer of the transformer's encoder. It encapsulates a multi-head self-attention mechanism followed by the position-wise feed-forward neural network, with residual connections, layer normalization, and dropout applied as appropriate. Together, these components allow the encoder to capture complex relationships in the input data and transform them into a useful representation for downstream tasks. Typically, multiple such encoder layers are stacked to form the complete encoder part of a transformer model.

**Output:** The output of the final encoder layer is a set of vectors, each representing the input sequence with a rich contextual understanding. This output is then used as the input for the decoder in a Transformer model.

This careful encoding paves the way for the decoder, guiding it to pay attention to the right words in the input when it's time to decode.

Think of it like building a tower, where you can stack up N encoder layers. Each layer in this stack gets a chance to explore and learn different facets of attention, much like layers of knowledge. This not only diversifies the understanding but could significantly amplify the predictive capabilities of the transformer network.

```python
class EncoderLayer(nn.Module):
    def __init__(self, d_model: int, num_heads: int, d_ff: int, dropout: float):
        """ 
        Encoder layer which encapstulates the attention mecanism and the ff NN to process the input embeddings.
        Parameter(s):
        d_model(int): The dimensionality of the input embeddings.
        num_heads(int): The number of attention heads in the multi-head attention.
        d_ff(int): The dimensionality of the inner layer in the PositionWise FFN.
        dropout(float): The drop out rate used for regularization.
        """
        super(EncoderLayer, self).__init__()
        self.attn = MultiHeadAttention(d_model, num_heads) #Multi-head attention mechanism.
        self.feed_forward = PositionWiseFeedForward(d_model, d_ff) #Position-wise feed-forward neural network.
        self.norm1 = nn.LayerNorm(d_model) # Layer normalization, applied to smooth the layer's input.
        self.norm2 = nn.LayerNorm(d_model) #
        self.dropout = nn.Dropout(dropout) # Dropout layer, used to prevent overfitting by randomly setting some activations to zero during training.

    def forward(self, x, mask):
        attn_output = self.attn(x, x, x, mask)
        x = self.norm1(x + self.dropout(attn_output)) #The attention output is added to the original input (residual connection to help mitigate the vanishing gradient problem, allowing deeper models), followed by dropout and normalization using norm1. 
        ff_output = self.feed_forward(x)
        x = self.norm2(x + self.dropout(ff_output))
        return x
```

#### Building the decoder blocks

![Decoder_architecture.png](Decoder_architecture.png)

**Architecture:** The DecoderLayer class defines a single layer of the transformer's decoder. It consists of a multi-head self-attention mechanism, a multi-head cross-attention mechanism (that attends to the encoder's output), a position-wise feed-forward neural network, and the corresponding residual connections, layer normalization, and dropout layers. This combination enables the decoder to generate meaningful outputs based on the encoder's representations, taking into account both the target sequence and the source sequence. As with the encoder, multiple decoder layers are typically stacked to form the complete decoder part of a transformer model.

**Output:** The final of the decoder's process involves a linear layer, serving as a classifier, topped off with a softmax function to calculate the probabilities of different words.

The Transformer decoder has a structure specifically designed to generate this output by decoding the encoded information step by step.

It is important to notice that the decoder operates in an autoregressive manner, kickstarting its process with a start token. It cleverly uses a list of previously generated outputs as its inputs, in tandem with the outputs from the encoder that are rich with attention information from the initial input.

This sequential dance of decoding continues until the decoder reaches a pivotal moment: the generation of a token that signals the end of its output creation.

```python
class DecoderLayer(nn.Module):
    def __init__(self, d_model: int, num_heads: int, d_ff: int, dropout: float):
        """ 
        Decoder layer that encapsulates the attention mecanism and feed forward NN to generate outputs based on the encoder's representations taking into account both the target sequence and the source sequence.
        Parameter(s):
        d_model(int): The dimensionality of the input embeddings.
        num_heads(int): The number of attention heads in the multi-head attention.
        d_ff(int): The dimensionality of the inner layer in the PositionWise FFN.
        dropout(float): The drop out rate used for regularization.
        """
        super(DecoderLayer, self).__init__()
        self.attn = MultiHeadAttention(d_model, num_heads) #Multi-head self-attention mechanism for the target sequence.
        self.cross_attn = MultiHeadAttention(d_model, num_heads) #Multi-head attention mechanism that attends to the encoder's output.
        self.feed_forward = PositionWiseFeedForward(d_model, d_ff)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.norm3 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, enc_output, src_mask, tgt_mask):
        attn_output = self.attn(x, x, x, tgt_mask) #The input x is processed through a masked self-attention mechanism -> it prevents positions from attending to subsequent positions, which means that each word in the sequence isn't influenced by future tokens.
        x = self.norm1(x + self.dropout(attn_output))
        cross_attn_output = self.cross_attn(x, enc_output, enc_output, src_mask) #The normalized output from the previous step is processed through a cross-attention mechanism that attends to the encoder's output enc_output. Here, the outputs from the encoder take on the roles of both queries and keys, while the outputs from the first multi-headed attention layer of the decoder serve as values. This setup effectively aligns the encoder's input with the decoder's, empowering the decoder to identify and emphasize the most relevant parts of the encoder's input.
        x = self.norm2(x + self.dropout(cross_attn_output))
        ff_output = self.feed_forward(x)
        x = self.norm3(x + self.dropout(ff_output))
        return x
```

#### Combining the encoder and decoder layers to create the complete Transformer network

![Transformer_architecture.png](Transformer_architecture.png)

The Transformer class brings together the various components of a Transformer model, including the embeddings, positional encoding, encoder layers, and decoder layers. It provides a convenient interface for training and inference, encapsulating the complexities of multi-head attention, feed-forward networks, and layer normalization.

This implementation follows the standard Transformer architecture, making it suitable for sequence-to-sequence tasks like machine translation, text summarization, etc. Including masking ensures that the model adheres to the causal dependencies within sequences, ignoring padding tokens and preventing information leakage from future tokens.

These sequential steps empower the Transformer model to efficiently process input sequences and produce corresponding output sequences.

Petite subtilité à noter

On remarque une asymétrie : src_mask masque les positions key correspondant au padding, alors que la partie padding de tgt_mask masque les positions query, pas les positions key de la séquence cible. Concrètement, ça veut dire que dans la self-attention du decoder, un token valide pourrait techniquement encore "regarder" un token <pad> de la séquence cible comme key (le mask causal l'empêchera seulement si ce padding est dans le futur, pas s'il est dans le passé — ce qui n'arrive normalement pas si le padding est bien à droite de la séquence, après le token <end>). C'est une implémentation très répandue dans les tutoriels, qui fonctionne correctement tant que le padding est toujours placé après la fin de séquence utile (convention standard), ce qui est le cas en pratique la plupart du temps.

```python
class Transformer(nn.Module):
    def __init__(self, src_vocab_size: int, tgt_vocab_size: int, d_model: int, num_heads: int, num_layers: int, d_ff: int, max_seq_length: int, dropout: float):
        """ 
        Returns the overall architecture of the transformer including the embeddings, positional encoding, encoder layers and decoder layers.
        Parameter(s):
        src_vocab_size(int): Source vocabulary size
        tgt_vocab_size(int): Target vocabulary size
        d_model(int): The dimensionality of the model's embeddings.
        num_heads(int): Number of attention heads in the multi-head attention mechanism
        num_layers(int): Number of layers for both the encoder and the decoder.
        d_ff(int): Dimensionality of the inner layer in the feed-forward network.
        max_seq_length(int): Maximum sequence length for positional encoding.
        dropout(float): Dropout rate for regularization.
        """
        super(Transformer, self).__init__()
        self.positional_encoding = PositionalEncoding(d_model, max_seq_length) #Positional encoding component.
        self.encoder_embedding = nn.Embedding(src_vocab_size, d_model) #Embedding layer for the source sequence.
        self.decoder_embedding = nn.Embedding(tgt_vocab_size, d_model) #Embedding layer for the target sequence.

        self.encoder_layers = nn.ModuleList([EncoderLayer(d_model, num_heads, d_ff, dropout) for _ in range(num_layers)]) #A list of encoder layers.
        self.decoder_layers = nn.ModuleList([DecoderLayer(d_model, num_heads, d_ff, dropout) for _ in range(num_layers)]) #A list of decoder layers.

        self.fc = nn.Linear(d_model, tgt_vocab_size) #Final fully connected (linear) layer mapping to target vocabulary size.
        self.dropout = nn.Dropout(dropout) #Dropout layer.

    #This method is used to create masks for the source and target sequences, ensuring that padding tokens are ignored and that future tokens are not visible during training for the target sequence.
    def generate_mask(self, src, tgt):
        """ 
        Effet concret du src_mask: pour n'importe quelle position query t et n'importe quelle tête, on masque les positions key s qui correspondent à du padding dans la séquence source — peu importe qui regarde, on empêche de prêter attention à un token <pad>, qui ne porte aucune information utile.
        C'est utilisé à deux endroits : dans la self-attention de l'encoder (les tokens sources ignorent le padding source entre eux), et dans la cross-attention du decoder (le decoder ignore le padding source quand il regarde la sortie de l'encoder).

        Effet concret du tgt_mask: Effet concret : ce mask dit "si le token query (celui qui pose la question) est lui-même un <pad>, on masque toute sa ligne de scores" — peu importe le token key regardé. 
        En pratique cette info compte peu, car on n'utilisera jamais la sortie calculée pour une position de padding (elle est ignorée plus tard, typiquement via la loss qui exclut les positions <pad>).

        Effet concret du masque causal (nopeak_mask): la ligne i (position query t=i) a des True uniquement pour les colonnes j ≤ i. Concrètement, le token en position t ne peut regarder que lui-même et les tokens précédents, jamais les tokens futurs. 
        Exactement le masque causal qui empêche le decoder de "voir" la suite de la phrase qu'il doit prédire.
        """
        src_mask = (src != 0).unsqueeze(1).unsqueeze(2) #src a une shape (batch, src_len) et contient les indices des tokens (0 étant typiquement l'indice du token <pad>). (src != 0) a la meme shape et est un booléen qui vaut True si le token est vrai token False si c'est du padding. unsqueeze(1).unsqueeze(2) a la shape (batch, 1, 1, src_len) pour matcher avec la shape de attn scores (batch, num_heads, seq_len_q, seq_len_k) et donc broadcaster sur num_head et seq_len_q (query). EX: 1 vs num_heads : le même masque est utilisé pour toutes les têtes.
        tgt_mask = (tgt !=0).unsqueeze(1).unsqueeze(3) #shape (batch, 1, tgt_len, 1) -> mask s'applique sur la dimension query, et broadcast sur la dimension key.
        seq_length = tgt.size(1)
        nopeak_mask = (1 - torch.triu(torch.ones(1, seq_length, seq_length, device=tgt.device), diagonal=1)).bool() #1 - torch.triu(torch.ones(1, seq_length, seq_length) donne la partie strictement en dessous de la diagonale mise à 1 et .bool() transforme en booléen: True = autorisé, False = Interdit. shape (1, seq_length, seq_length)
        tgt_mask = tgt_mask & nopeak_mask #Combinaison finale: tgt_mask (padding) : (batch, 1, tgt_len, 1) nopeak_mask (causal) : (1, tgt_len, tgt_len). Le & (AND logique élément par élément) broadcast les deux shapes ensemble : (batch, 1, tgt_len, 1)  &  (1, 1, tgt_len, tgt_len)  →  (batch, 1, tgt_len, tgt_len). Résultat : pour chaque paire (t, s), l'attention est autorisée seulement si :le token query t n'est pas du padding, ET s ≤ t (pas de triche vers le futur).
        return src_mask, tgt_mask
    
    def forward(self, src, tgt):
        src_mask, tgt_mask = self.generate_mask(src, tgt)
        src_embedded = self.dropout(self.positional_encoding(self.encoder_embedding(src)))
        tgt_embedded = self.dropout(self.positional_encoding(self.decoder_embedding(tgt)))

        enc_output = src_embedded
        for enc_layer in self.encoder_layers:
            enc_output = enc_layer(enc_output, src_mask)

        dec_output = tgt_embedded
        for dec_layer in self.decoder_layers:
            dec_output = dec_layer(dec_output, enc_output, src_mask, tgt_mask)

        output = self.fc(dec_output)

        return output
```

#### Data preparation & Training loop

```python
import math
import os

import sentencepiece as spm #tokenizer to be trained on multi30k ds
import torch
import torch.nn as nn
from datasets import load_dataset
from torch.nn.utils.rnn import pad_sequence
from torch.utils.data import DataLoader, Dataset

# ------------------------------------------------------------------
# Configuration
# ------------------------------------------------------------------
VOCAB_SIZE = 8000
MAX_LEN = 64            # doit être <= max_seq_length du modèle
D_MODEL, NUM_HEADS, NUM_LAYERS, D_FF, DROPOUT = 256, 8, 3, 1024, 0.1
BATCH_SIZE = 64
EPOCHS = 15
LR, WARMUP = 5e-4, 800
SRC_COL, TGT_COL = "en", "fr"   # à vérifier avec print(ds["train"][0])
device = "cuda" if torch.cuda.is_available() else "mps" if torch.mps.is_available() else "cpu"


# ------------------------------------------------------------------
# 1) Données
# ------------------------------------------------------------------
ds = load_dataset("ryan82/multi30k_fr")
print(ds)
print(ds["train"][0])

# Split de validation : on utilise celui du dataset s'il existe
train_split = ds["train"]
val_name = "validation" if "validation" in ds else ("val" if "val" in ds else None)
if val_name is None:
    tmp = train_split.train_test_split(test_size=0.05, seed=0)
    train_split, val_split = tmp["train"], tmp["test"]
else:
    val_split = ds[val_name]

# ------------------------------------------------------------------
# 2) Tokenizer (entraîné sur EN + FR, pad=0 indispensable pour generate_mask)
# ------------------------------------------------------------------
if not os.path.exists("multi30k_bpe.model"):
    with open("corpus.txt", "w", encoding="utf-8") as f:
        for ex in train_split:
            f.write(ex[SRC_COL].strip() + "\n")
            f.write(ex[TGT_COL].strip() + "\n")
    spm.SentencePieceTrainer.train(
        input="corpus.txt", model_prefix="multi30k_bpe",
        vocab_size=VOCAB_SIZE, model_type="bpe",
        pad_id=0, bos_id=1, eos_id=2, unk_id=3,
        character_coverage=1.0,
    )

sp = spm.SentencePieceProcessor(model_file="multi30k_bpe.model")
PAD, BOS, EOS = sp.pad_id(), sp.bos_id(), sp.eos_id()
assert (PAD, BOS, EOS) == (0, 1, 2)
vocab_size = sp.get_piece_size() #size of vocab = 8000 here


class PairDataset(Dataset):
    def __init__(self, split):
        self.pairs = []
        for ex in split:
            s = sp.encode(ex[SRC_COL].strip())
            t = sp.encode(ex[TGT_COL].strip())
            # on garde une place pour BOS/EOS
            if len(s) <= MAX_LEN - 1 and len(t) <= MAX_LEN - 2:
                self.pairs.append((s, t))

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, i):
        return self.pairs[i]


def collate(batch):
    src = [torch.tensor(s + [EOS]) for s, t in batch]
    tgt_in = [torch.tensor([BOS] + t) for s, t in batch]
    tgt_out = [torch.tensor(t + [EOS]) for s, t in batch]
    pad = lambda x: pad_sequence(x, batch_first=True, padding_value=PAD)
    return pad(src), pad(tgt_in), pad(tgt_out)


train_loader = DataLoader(PairDataset(train_split), batch_size=BATCH_SIZE,
                          shuffle=True, collate_fn=collate)
val_loader = DataLoader(PairDataset(val_split), batch_size=BATCH_SIZE,
                        shuffle=False, collate_fn=collate)

# ------------------------------------------------------------------
# 3) Modèle, loss, optimiseur
# ------------------------------------------------------------------
model = Transformer(vocab_size, vocab_size, D_MODEL, NUM_HEADS, NUM_LAYERS,
                    D_FF, MAX_LEN, DROPOUT).to(device)
print(f"Paramètres : {sum(p.numel() for p in model.parameters()) / 1e6:.1f} M")


criterion = nn.CrossEntropyLoss(ignore_index=PAD, label_smoothing=0.1)
optimizer = torch.optim.Adam(model.parameters(), lr=LR, betas=(0.9, 0.98), eps=1e-9)
# warmup linéaire puis décroissance en 1/sqrt(step)
scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lambda step: min((step + 1) / WARMUP, math.sqrt(WARMUP / (step + 1)))
)


def run_epoch(loader, train):
    model.train(train)
    total, n = 0.0, 0
    with torch.set_grad_enabled(train):
        for src, tgt_in, tgt_out in loader:
            src, tgt_in, tgt_out = src.to(device), tgt_in.to(device), tgt_out.to(device)
            logits = model(src, tgt_in)  # (batch, tgt_len, vocab)
            loss = criterion(logits.reshape(-1, logits.size(-1)), tgt_out.reshape(-1))
            if train:
                optimizer.zero_grad()
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
                scheduler.step()
            total += loss.item()
            n += 1
    return total / n


# ------------------------------------------------------------------
# 4) Traduction (greedy)
# ------------------------------------------------------------------
@torch.no_grad()
def translate(sentence, max_new_tokens=MAX_LEN - 2):
    model.eval()
    src = torch.tensor([sp.encode(sentence)[:MAX_LEN - 1] + [EOS]], device=device)
    tgt = torch.tensor([[BOS]], device=device)
    for _ in range(max_new_tokens):
        logits = model(src, tgt)
        next_tok = logits[:, -1].argmax(dim=-1, keepdim=True)
        if next_tok.item() == EOS:
            break
        tgt = torch.cat([tgt, next_tok], dim=1)
    return sp.decode(tgt[0, 1:].tolist())


# ------------------------------------------------------------------
# 5) Entraînement
# ------------------------------------------------------------------
if __name__ == "__main__":
    best_val = float("inf")
    test_sentences = [
        "A man is sleeping on the couch.",
        "Two children are playing in the park.",
        "A woman is riding a bike near the beach.",
    ]
    for epoch in range(1, EPOCHS + 1):
        tr = run_epoch(train_loader, train=True)
        va = run_epoch(val_loader, train=False)
        print(f"Epoch {epoch:02d} | train loss {tr:.3f} | val loss {va:.3f}")
        if va < best_val:
            best_val = va
            torch.save(model.state_dict(), "best_model.pt")
        for s in test_sentences:
            print(f"   {s}  ->  {translate(s)}")
```
