```python
def bubble_sort(L:list[float])->list:

    """
    Sorts a list in ascending order using bubble sort
    Parameter(s):
    L (list[float]): Input list
    Return(s):
    list: Sorted list
    """

    if not isinstance(L, list):
        raise TypeError("Input should be a list")
    
    bridge = 0

    while not all(L[i]<L[i+1] for i in range(len(L)-1)):
        for i in range(len(L)-1):
            if L[i]>L[i+1]:
                bridge = L[i]
                L[i] = L[i+1]
                L[i+1] = bridge
                
    return L


def bubble_sort(L: list[float]) -> list:
    """
    Sorts a list in ascending order using bubble sort
    Parameter(s):
    L (list[float]): Input list
    Return(s):
    list: Sorted list
    """
    if not isinstance(L, list):
        raise TypeError("Input should be a list")
    
    n = len(L)
    for i in range(n):
        swapped = False
        for j in range(n - 1 - i):
            if L[j] > L[j+1]:
                L[j], L[j+1] = L[j+1], L[j]
                swapped = True
        if not swapped:
            break  # déjà trié, on arrête
    return L


bubble_sort([5,2,4,1])
```

```python
def string_compression(S:str)->str:
    """ 
    Returns the compressed format of a string.

    Parameter(s):
    S (str): Input string

    Return(s):
    Returning the compressed format of the input string
    """

    if not isinstance(S, str):
        raise TypeError("Input should be a string")
    
    format_str = ''
    current_char = ''

    for i,s in enumerate(S):
        if not current_char:
            current_char = s
            count = 1
            continue
        if s == current_char:
            count += 1
        if s != current_char:
            format_str += current_char + f'{count}'
            current_char = s
            count = 1
        
        if i == len(S)-1:
            format_str += current_char + f'{count}'
            return format_str
        
```

```python
def factor_recursive(n:int)->int:
    """ 
    Returns the factorial of an integer

    Parameter(s):
    n (int): Input integer

    Return(s):
    Returning the factorial of the input integer
    """

    if not isinstance(n, int) or not n > 0:
        raise TypeError("Input should a positive integer")
    

    if n ==1 or n ==2:
        facto = n
    else:
        facto = n * factor_recursive(n-1)

    return facto
```

```python
import numpy as np

def gradient_descent(X: list, y: float, epochs: int = 1000, lr: float = 1e-1):
    """
    Trains a linear regression model via gradient descent
    Parameter(s):
    X (list): Input features
    y (float): Target value
    epochs (int): Number of training iterations
    lr (float): Learning rate
    Return(s):
    tuple: Trained weights W and bias b
    """
    if not isinstance(X, list):
        raise TypeError("Input values should be a list")
    if not isinstance(y, float):
        raise TypeError("Target value should be a float")
    
    # Initialisation
    W = [np.random.uniform(0., 1.) for _ in X]
    b = 0.0
    
    for epoch in range(epochs):
        # Forward pass
        y_pred = b
        for x, w in zip(X, W):
            y_pred += x * w
        
        # Loss (MSE)
        L = (y_pred - y) ** 2
        
        # Backpropagation
        dL_dW = [2 * (y_pred - y) * x for x in X]
        dL_db = 2 * (y_pred - y)
        
        # Update
        W = [w - lr * dw for w, dw in zip(W, dL_dW)]
        b = b - lr * dL_db
        
        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {L:.4f}")
    
    return (W, b)

gradient_descent([2,3], 1.)
```

```python
import math as m
def logistic_regression(X: list, y: float, epochs: int = 1000, lr: float = 1e-1):
    """ 
    Trains a logistic regression by gradient descent

    Parameter(s):
    X (list): Input features
    y (float): Target value
    epochs (int): Number of training iterations
    lr (float): Learning rate

    Return(s):
    tuple: Updates weights and bias.
    """

    if not isinstance(X, list) or not isinstance(y, float):
        raise TypeError("Input features should be given as a list, and the target value should be a float")
    
    ## Initialization
    W = [np.random.uniform(0., 1.) for x in X]
    b = 0.0

    ## Sigmoid function
    def sigmoid(z):
        return 1/(1 + m.exp(-z))
    
    
    for epoch in range(epochs):
        # Forward pass + sigmoid
        z = b
        for x,w in zip(X,W):
            z += x * w
            #y_pred = sigmoid(z)
            y_pred = min(max(sigmoid(z), 1e-7), 1 - 1e-7)  # clipping

        # Loss (Cross-entropy)
        CE = -(y * m.log(y_pred) + (1-y) * m.log(1-y_pred))

        # Back propagation
        dL_dW = [(y_pred - y) * x for x in X]
        dL_db = (y_pred - y)

        # Update
        W = [w - lr * dw for (w,dw) in zip(W, dL_dW)]
        b = b - lr * dL_db

        if epoch % 100 == 0:
            print(f"Epoch: {epoch}, Loss: {CE}")

    return (W,b)


logistic_regression([2.71, 3.33], 0.86)
```

```python
def sample_split(X: np.ndarray, y: np.ndarray, 
                 train_size: float = 0.8, 
                 test_size: float = 0.1, 
                 val_size: float = 0.1):
    """
    Returns training, test and validation splits from input data
    Parameter(s):
    X (np.ndarray): Input features
    y (np.ndarray): Labels
    train_size (float): Proportion for training
    test_size (float): Proportion for test
    val_size (float): Proportion for validation
    Return(s):
    tuple: X_train, y_train, X_test, y_test, X_val, y_val
    """
    if not isinstance(X, np.ndarray) or not isinstance(y, np.ndarray):
        raise TypeError("Features and labels should be numpy arrays")
    if not all(isinstance(s, float) for s in [train_size, test_size, val_size]):
        raise ValueError("Split sizes should be floats")
    if not round(train_size + test_size + val_size, 5) == 1.0:
        raise ValueError("Split sizes should sum to 1.0")
    
    n = X.shape[0]
    
    # Shuffle des indices
    indices = np.random.permutation(n)
    
    # Calcul des seuils
    train_end = round(train_size * n)
    test_end  = train_end + round(test_size * n)
    
    train_idx = indices[:train_end]
    test_idx  = indices[train_end:test_end]
    val_idx   = indices[test_end:]
    
    return (X[train_idx], y[train_idx],
            X[test_idx],  y[test_idx],
            X[val_idx],   y[val_idx])

X = np.array([[2,3,7], [3,4,8], [6,8,9], [5,9,1], [8,7,6], [9,8,1]])
y = np.array([0,1,1,2,0,1])
sample_split(X, y, train_size=0.7, test_size=0.2, val_size=0.1)
```

```python
import math as m
def metrics(y_pred : np.ndarray, y_true : np.ndarray, method : str = "regression"):

    """
    Parameter(s):
    y_pred (np.ndarray): Predictions from the model
    y_true (np.ndarray): True values (or labels for classif)
    method (str): Type of problem solved by the model, either 'regression' or 'classification'

    Return(s):
    tuple: MSE, MAE, RMSE, accuracy, F1
    """

    if not isinstance(y_pred, np.ndarray) or not isinstance(y_true, np.ndarray):
        raise TypeError("Value inputs should be arrays")
    if not isinstance(method, str) or method not in ["regression", "classification"]:
        raise ValueError("Method should be a string either 'regression' or 'classification'")
    if not len(y_pred)==len(y_true):
        raise ValueError("Input arrays should be the same size")
    
    n = len(y_pred)

    ### MSE, MAE and RMSE for regression
    if method == "regression":
        MSE = (1/n) * sum((y_pred-y_true)**2)
        MAE = (1/n) * sum(abs(y_pred-y_true))
        RMSE = m.sqrt(MSE)
        return MSE, MAE, RMSE
    

    ### Accuracy, precision, recall and F1 score
    else:
        TP = np.count_nonzero((y_true==1) & (y_pred==1))
        TN = np.count_nonzero((y_true==0) & (y_pred==0))
        FP = np.count_nonzero((y_true==0) & (y_pred==1))
        FN = np.count_nonzero((y_true==1) & (y_pred==0))

        accuracy = (TP+TN) / (TP+TN+FP+FN)

        # Si TP+FP == 0 ou TP+FN == 0 → division par zéro
        precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
        recall    = TP / (TP + FN) if (TP + FN) > 0 else 0.0

        F1 = 2 * (precision*recall) / (precision+recall) if (precision+recall) > 0 else 0.0

                
        return accuracy, precision, recall, F1


y_true_regression = np.array([1.2, 3.2, 4.5, 9.3])
y_pred_regression = np.array([1.1, 3.5, 4.6, 8.8])

y_true = np.array([0, 1, 1, 0])
y_pred = np.array([1, 1, 0, 0])

metrics(y_true, y_pred, 'classification')
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class CNN_images(nn.Module):
    def __init__(self, in_channels:int, num_classes:int):
        """
        CNN for image classification (e.g. CIFAR10)
        Parameter(s):
        in_channels (int): Number of input channels (3 for RGB)
        num_classes (int): Number of output classes
        """
        super().__init__()
        #1st convolution layer
        self.conv2d_1 = nn.Conv2d(in_channels=in_channels, out_channels=16, kernel_size=3, padding=1)
        #2nd convolution layer
        self.conv2d_2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)

        #1 maxpool layer
        self.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Adptative pool to adapt to different images sizes
        self.adaptive_pool = nn.AdaptiveAvgPool2d((8, 8))  # force toujours 8×8

        #2nd maxpool layer
        self.linear = nn.Linear(32*8*8, num_classes)

    def forward(self, x):

        """
        Define the forward pass of the neural network.

        Parameters:
           x (torch.tensor): Input tensor.

        Returns:
            torch.Tensor
                The output tensor after passing through the network.
       """

        x = F.relu(self.conv2d_1(x))
        x = self.maxpool(x)
        x = F.relu(self.conv2d_2(x))
        x = self.maxpool(x)
        x = self.adaptive_pool(x)
        x = x.reshape(x.shape[0], -1)
        x = self.linear(x)

        return x

class CNN_spectra(nn.Module):
    def __init__(self, in_channels:int, num_classes:int):
        """
        1D CNN for spectra classification (e.g. FSRQ / BL Lacs / BCUs)
        Parameter(s):
        in_channels (int): Number of input channels (1 for single flux array)
        num_classes (int): Number of output classes
        """
        super().__init__()
        #1st convolution layer
        self.conv1d_1 = nn.Conv1d(in_channels=in_channels, out_channels=8, kernel_size=3, padding=1)
        #2nd convolution layer
        self.conv1d_2 = nn.Conv1d(in_channels=8, out_channels=16, kernel_size=3, padding=1)

        #1 maxpool layer
        self.maxpool = nn.MaxPool1d(kernel_size=2, stride=2)

        # Adaptative pool to adapt to any spectrum length
        self.adaptive_pool = nn.AdaptiveAvgPool1d(32)  # robuste à toute longueur de spectre

        #2nd maxpool layer
        self.linear = nn.Linear(16*32, num_classes)

    def forward(self, x):

        x = F.relu(self.conv1d_1(x))
        x = self.maxpool(x)
        x = F.relu(self.conv1d_2(x))
        x = self.maxpool(x)
        x = self.adaptive_pool(x)
        x = x.reshape(x.shape[0], -1)
        x = self.linear(x)

        return x

device = "cuda" if torch.cuda.is_available() else "mps" #or 'mps' for macOS
modelA = CNN_images(in_channels=3, num_classes=10).to(device)
print(modelA)

modelB = CNN_spectra(in_channels=1, num_classes=3).to(device) #ex FSRQ, BL Lacs and BCUs
print(modelB)
```

```python
from torch.utils.data import DataLoader, TensorDataset
import random

class Stacked_LSTM(nn.Module):
    def __init__(self, input_size: int, num_hidden: int = 64, num_layers: int = 2):
        """
        Stacked LSTM for light-curve binary classification (e.g. FSRQ vs BL Lac)
        Parameter(s):
        input_size (int): Number of features per timestep
        num_hidden (int): Hidden state size
        num_layers (int): Number of stacked LSTM layers
        """
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=num_hidden,
            num_layers=num_layers,
            batch_first=True,
            dropout=0.2
        )
        self.linear = nn.Linear(num_hidden, 1)  # binaire → 1 sortie

    def forward(self, x):
        output, (h_n, c_n) = self.lstm(x)
        x = self.linear(h_n[-1])  # dernier hidden state → Linear
        return x


def create_LCs(num_LC: int = 100, seq_len: int = 64):
    """
    Creates fictive light-curves with binary labels
    Parameter(s):
    num_LC (int): Number of light-curves
    seq_len (int): Length of each light-curve
    Return(s):
    tuple: X_tensor (num_LC, seq_len, 1), y_tensor (num_LC,)
    """
    X = np.random.uniform(0, 1, size=(num_LC, seq_len, 1))  # (batch, seq, features)
    y = np.random.randint(0, 2, size=(num_LC,))

    X_tensor = torch.tensor(X, dtype=torch.float32)
    y_tensor = torch.tensor(y, dtype=torch.float32)

    return X_tensor, y_tensor


# Setup
device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"

X_tensor, y_tensor = create_LCs()
loader = DataLoader(TensorDataset(X_tensor, y_tensor), batch_size=32, shuffle=True)

# Modèle, loss, optimizer
model_LSTM = Stacked_LSTM(input_size=1).to(device)
loss_fn = nn.BCEWithLogitsLoss()  # sigmoid + BCE — plus stable
optimizer = torch.optim.Adam(model_LSTM.parameters(), lr=1e-3)

# Boucle d'entraînement
loss_history = []
for epoch in range(10):
    model_LSTM.train()  # mode train (active dropout)
    epoch_loss = 0.0

    for X_batch, y_batch in loader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        output = model_LSTM(X_batch).squeeze(-1)  # (batch, 1) → (batch,)
        loss = loss_fn(output, y_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(loader)
    loss_history.append(avg_loss)
    print(f"Epoch {epoch+1}/10 — Loss: {avg_loss:.4f}")

# Classification
model_LSTM.eval()
with torch.no_grad():
    X_tensor = X_tensor.to(device)
    logits = model_LSTM(X_tensor).squeeze(-1)   # scores bruts
    probs = torch.sigmoid(logits)                # → probabilités
    preds = (probs > 0.5).int()                  # → classes 0 ou 1
    print(f"Predictions: {preds[:10]}")
    print(f"True labels: {y_tensor[:10].int()}")
```

```python
class AutoEncoder(nn.Module):
    def __init__(self, input_size: int):
        """ Docstring.."""
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(input_size, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
        )

        self.decoder = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, input_size),
            nn.Sigmoid()
        )

    def forward(self, x):

        latent = self.encoder(x)
        reconst = self.decoder(latent)
        return reconst


device = "cuda" if torch.cuda.is_available() else "mps" if torch.mps.is_available() else "cpu"
model_AE = AutoEncoder(784).to(device)
optimizer = torch.optim.SGD(model_AE.parameters(), lr = 0.001, momentum=0.1)
loss_fn = nn.MSELoss()
print(model_AE)


def training_one_epoch(model, train_loader, opti, logging_interval=100):
    running_loss = 0

    for batch_idx, (img, label) in enumerate(train_loader):
        
        # move features to mps to be coherent with the model
        img = img.reshape(-1, 28*28).to(device)

        #Make reconstruction for this batch
        recon_img = model(img)

        # Zero the gradient for every beatch
        opti.zero_grad()

        # Compute the loss and its gradient
        loss = loss_fn(recon_img, img)
        loss.backward()

        # Adjust lr
        optimizer.step()

        running_loss += loss.item()

        # LOGGING (Gather data and report)
        #if not batch_idx % logging_interval:
        #    print(f'Batch: {batch_idx}/{len(training_loader)} | Loss: {loss}')

    avg_loss = running_loss/len(training_loader)

    return avg_loss

#avg_loss = training_one_epoch(model_AE, training_loader, optimizer)


def train_AE(EPOCH, model, optimizer,
             train_loader, validation_loader, loss_fn=loss_fn, epoch_number = 0):

    log_dict = {'train_loss_per_batch':[],
                    'train_loss_per_epoch':[]}

    running_vloss = 0

    for epoch in range(EPOCH):
        print(f'EPOCH {epoch_number+1}:')

        ## Training
        model.train()
        avg_loss = training_one_epoch(model, training_loader, optimizer)

        # Set the model to evaluation mode, disabling dropout and using population
        # statistics for batch normalization.
        model_AE.eval()
        
        # Disable gradient computation and reduce memory consuption.
        with torch.no_grad():
            for batch_idx, (vimg, vlabel) in  enumerate(validation_loader):
                vimg = vimg.reshape(-1,28*28).to(device)
                vrecon = model(vimg)
                vloss = loss_fn(vrecon, vimg)
                running_vloss += vloss

        avg_vloss = running_vloss / len(validation_loader)
        print(f'train loss: {avg_loss} | val loss: {avg_vloss}')

        epoch_number += 1


train_AE(5, model_AE, optimizer, training_loader, validation_loader)
```
